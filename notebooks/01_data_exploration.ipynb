# 01 - Data Exploration

Exploratory data analysis of NBA player box score data.

```python
import sys
sys.path.insert(0, '..')

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from src.config.settings import Config
from src.utils.io import load_parquet

sns.set_theme(style='whitegrid')
%matplotlib inline
```

```python
# Load cleaned data
df = load_parquet('../data/processed/cleaned_box_scores.parquet')
print(f'Loaded {len(df)} records')
print(f'Seasons: {df["SEASON"].min()} - {df["SEASON"].max()}')
print(f'Players: {df["PLAYER_ID"].nunique()}')
df.head()
```

```python
# Distribution of target stats
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
for ax, col, title in zip(axes.flat, ['PTS', 'REB', 'FG3M', 'FTM'],
                          ['Points', 'Rebounds', '3-Pointers Made', 'Free Throws Made']):
    ax.hist(df[col].dropna(), bins=50, alpha=0.7, edgecolor='black')
    ax.set_title(title)
    ax.set_xlabel('Value')
    ax.set_ylabel('Frequency')
    ax.axvline(df[col].mean(), color='red', linestyle='--', label=f'Mean: {df[col].mean():.1f}')
    ax.legend()
plt.tight_layout()
plt.show()
```

```python
# Games per player distribution
games_per_player = df.groupby('PLAYER_ID')['GAME_ID'].nunique()
print(f'Games per player: mean={games_per_player.mean():.0f}, median={games_per_player.median():.0f}')
games_per_player.hist(bins=50)
plt.title('Games per Player')
plt.xlabel('Number of Games')
plt.ylabel('Number of Players')
plt.show()
```
