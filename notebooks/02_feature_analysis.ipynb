# 02 - Feature Analysis

Analysis of engineered features for NBA player prop prediction.

```python
import sys
sys.path.insert(0, '..')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from src.utils.io import load_parquet
from src.data.features import FeatureEngineer

sns.set_theme(style='whitegrid')
%matplotlib inline
```

```python
# Load feature-engineered data
df = load_parquet('../data/features/feature_engineered.parquet')
feature_cols = FeatureEngineer.get_feature_columns(df)
print(f'Number of features: {len(feature_cols)}')
print(f'Features: {feature_cols}')
```

```python
# Correlation between rolling features and targets
target_cols = ['PTS', 'REB', 'FG3M', 'FTM']
rolling_cols = [c for c in feature_cols if 'rolling' in c]

if rolling_cols:
    corr = df[rolling_cols + target_cols].corr().loc[rolling_cols, target_cols]
    plt.figure(figsize=(10, 8))
    sns.heatmap(corr, annot=True, cmap='RdBu_r', center=0, fmt='.2f')
    plt.title('Feature-Target Correlations')
    plt.tight_layout()
    plt.show()
```

```python
# Feature distributions
fig, axes = plt.subplots(3, 4, figsize=(16, 12))
for ax, col in zip(axes.flat, feature_cols[:12]):
    values = df[col].dropna()
    ax.hist(values, bins=50, alpha=0.7)
    ax.set_title(col, fontsize=9)
plt.tight_layout()
plt.show()
```

```python
# Missing value analysis
missing = df[feature_cols].isnull().sum().sort_values(ascending=False)
missing_pct = (missing / len(df) * 100).round(1)
print('Missing values (% of total):')
print(missing_pct[missing_pct > 0])
```
