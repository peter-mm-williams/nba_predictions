# 03 - Model Comparison

Compare prediction quality across model types.

```python
import sys
sys.path.insert(0, '..')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from src.utils.io import load_json
from src.evaluation.plots import (
    plot_model_comparison,
    plot_calibration,
    plot_distribution_comparison,
)

sns.set_theme(style='whitegrid')
%matplotlib inline
```

```python
# Load evaluation metrics for each model
models = {}
for model_name in ['xgboost', 'logistic', 'lstm', 'transformer']:
    try:
        metrics = load_json(f'../outputs/reports/metrics_{model_name}_points.json')
        models[model_name] = metrics
        print(f'{model_name}: {metrics}')
    except FileNotFoundError:
        print(f'{model_name}: not found')

if models:
    plot_model_comparison(models)
    plt.show()
```

```python
# Compare metrics across target stats
stats = ['points', 'rebounds', 'three_pointers_made', 'free_throws_made']
results = []
for stat in stats:
    for model_name in ['xgboost', 'logistic']:
        try:
            metrics = load_json(f'../outputs/reports/metrics_{model_name}_{stat}.json')
            results.append({'model': model_name, 'stat': stat, **metrics})
        except FileNotFoundError:
            pass

if results:
    results_df = pd.DataFrame(results)
    print(results_df.to_string(index=False))
```
